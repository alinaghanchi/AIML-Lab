```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Student": ["A","B","C","D","E","F","G","H","I","J","K","L","M","N","O"],
    "Study_Hours": [2,3,4,5,6,7,8,9,3,4,6,7,8,2,5],
    "Attendance": [65,70,75,80,82,85,90,92,68,72,78,88,91,60,76],
    "Marks": [45,50,58,65,70,75,82,88,48,55,68,78,85,40,62],
    "Assignments": [5,6,7,8,8,9,10,10,5,6,8,9,10,4,7],
    "Gender": ["F","M","F","M","F","M","F","M","F","M","F","M","F","M","F"]
}

df = pd.DataFrame(data)

df
```

```python
plt.plot(df["Student"], df["Marks"], marker='o')

plt.title("Student vs Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.grid(True)
plt.show()
```

```python
plt.plot(df["Student"], df["Marks"], linestyle= '--', marker='o')

plt.title("Student vs Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.grid(True)
plt.show()
```

```python
plt.plot(df["Student"], df["Marks"], marker='o', label="Marks")

plt.plot(df["Student"], df["Attendance"], marker='s', label="Attendance")

plt.title("Attendance vs Marks")
plt.xlabel("Students")
plt.ylabel("Values")

plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.scatter(
    df["Study_Hours"],
    df["Marks"]
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df["Student"], df["Marks"],
         marker='o',
         label='Marks')

plt.plot(df["Student"], df["Attendance"],
         marker='s',
         label='Attendance')

plt.plot(df["Student"], df["Study_Hours"],
         marker='^',
         label='Study Hours')

plt.title("Comparison of Marks, Attendance and Study Hours")
plt.xlabel("Students")
plt.ylabel("Values")

plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.hist(df["Marks"],
         bins=5,
         edgecolor='black')

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.scatter(df["Study_Hours"],
            df["Marks"])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.grid(True)
plt.show()
```

```python
plt.bar(df["Student"],
        df["Marks"])

plt.title("Marks of Students")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
gender_count = df["Gender"].value_counts()

plt.pie(gender_count,
        labels=gender_count.index,
        autopct='%1.1f%%')

plt.title("Gender Distribution")

plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)

plt.plot(df["Student"],
         df["Marks"],
         '--o')

plt.title("Marks")

plt.subplot(1,3,2)

plt.scatter(df["Study_Hours"],
            df["Marks"])

plt.title("Study Hours vs Marks")

plt.subplot(1,3,3)

gender_count = df["Gender"].value_counts()

plt.pie(gender_count,
        labels=gender_count.index,
        autopct='%1.1f%%')

plt.title("Gender")

plt.tight_layout()
plt.show()
```

```python
sns.barplot(
    x="Student",
    y="Marks",
    data=df
)

plt.title("Marks by Student")

plt.show()
```
